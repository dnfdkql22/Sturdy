```python
import pandas as pd
import numpy as np

# 1-1
ok = False

try:
  df_house = pd.read_csv('/content/1.01.house.price.csv')
except FileNotFoundError:
  print("파일명과 업로드 여부를 확인하세요.")
else:
  ok = True
  print(df_house.iloc[0])
  print(len(df_house))
  print(type(df_house['price'][0]))

try:
  df_test = pd.read_csv('/content/1.01.house.price_x.csv')
except FileNotFoundError:
  print("파일명과 업로드 여부를 확인하세요.")
else:
  print(df_test.iloc[0])
  print(len(df_test))
  print(type(df_test['price'][0]))
```

```python
# 1-2
if ok:
  houses = []
  count = 0
  count_err = 0
  for idx, row in df_house.iterrows():
    if pd.isna(row["ID"]) or pd.isna(row['size']) or pd.isna(row['year']) or pd.isna(row['price']):
      print(idx, "번 행 제외 : 필수 값 누락")
      count_err += 1
      continue
    try:
      houses.append({
          "ID": int(row['ID']),
          "size": int(row['size']),
          "year": int(row['year']),
          "price": float(row['price'])
      })
      count += 1
    except (ValueError, TypeError):
      print(idx, "번 행 제외 : 숫자로 바꿀 수 없음")
      count_err += 1
  print("정상 저장 행 수 :", count)
  print("제외 행 수 :", count_err)
```

```python
# 1-3
houses_test = houses.copy()
houses_test.append({"ID": 101, "year": 5, "price": 3})
houses_test.append({"ID": 102, "size": 35, "year": 5, "price": None})
houses_test.append({"ID": 103, "size": "오류", "year": 5, "price": 3})
houses_test.append({"ID": 104, "size": 35, "year": 5, "price": 3})


selected = []
for house in houses_test:
  if "size" not in house or "price" not in house:
    print(house["ID"], "제외 : size 또는 price 키 없음")
    continue
  if house["size"] is None or house["price"] is None or pd.isna(house["size"]) or pd.isna(house["price"]):
    print(house["ID"], "제외 : 값 없음(None 또는 결측값)")
    continue
  try:
    size = float(house["size"])
    price = float(house["price"])
  except (ValueError, TypeError):
    print(house["ID"], "제외 : 숫자로 바꿀 수 없음")
    continue
  if size <= 0 or price < 0:
    print(house["ID"], "제외 : size가 0 이하 또는 price가 음수")
    continue
  if size >= 30 and price <= 4:
    selected.append(house)
for house in selected:
  print(house["ID"])
print("선택한 주택 수 :", len(selected))
```

```python
# 1-4
def unit_price(price, size):
  try:
    price = float(price)
    size = float(size)
    if size <= 0:
      print("크기 오류: 크기는 0보다 커야 합니다.")
      return None
    if price < 0:
      print("가격 오류: 가격은 음수일 수 없습니다.")
      return None
    return price / size
  except ValueError:
    print("값 오류: 숫자로 변환할 수 없는 문자열입니다.")
    return None
  except TypeError:
    print("자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.")
    return None
for house in houses:
  result = unit_price(house["price"], house["size"])
  if result is not None:
    result = round(result, 3)
  print(house["ID"], result)

print(unit_price(4, 40))
print(unit_price("4", "40"))
print(unit_price(4, 0))
print(unit_price(-4, 40))
print(unit_price("오류", 40))
print(unit_price(None, 40))
print(unit_price(6, 30))
```
